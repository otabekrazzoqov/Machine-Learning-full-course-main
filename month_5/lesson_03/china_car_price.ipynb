```python
import pandas as pd

df = pd.DataFrame(all_cars)
df.to_csv("cars.csv", index=False)
print("Saved to cars.csv!")
```

## ***Scraping full dataset***

```python
import time
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC



# Keeping Chrome Browser open after program finishes

chrome_options = webdriver.ChromeOptions()
chrome_options.add_experimental_option("detach", True)
driver = webdriver.Chrome(options=chrome_options)

driver.get('https://cncargotop.com/?gad_source=1&gad_campaignid=23969307269&gbraid=0AAAABDyQWIBOsV3hcmFGl4s1jQHs69VIJ&gclid=CjwKCAjwx7LSBhB3EiwAjcodxPqSvms02WlC-5LD3nMJFDUX8H7nltdwqqPFlfRMDqU6vJZmimpqXRoCWgoQAvD_BwE#/home')
more = driver.find_element(By.CLASS_NAME, "more-lang")
more.click()


wait = WebDriverWait(driver, 10)
all_cars = []

def get_car_details(card):
    car = {}
    try:
        car["name"] = card.find_element(By.CLASS_NAME, "title").text
    except:
        car["name"] = None

    try:
        car["price"] = card.find_element(By.CLASS_NAME, "card_right_info").text
    except:
        car["price"] = None

    details = {}
    try:
        items = card.find_elements(By.CLASS_NAME, "item_info")
        for item in items:
            try:
                key = item.find_element(By.CLASS_NAME, "describe_name").text.strip()
                val = item.find_element(By.CLASS_NAME, "describe_value").text.strip()
                details[key] = val
            except:
                continue
    except:
        pass


    # mapping each field

    car["registration_year"]    = details.get("Registration Year", None)
    car["mileage_km"]           = details.get("Mileage(km)", None)
    car["fuel_type"]            = details.get("Fuel", None)
    car["engine_cc"]            = details.get("Engine(CC)", None)
    car["transmission"]         = details.get("Transmission", None)
    car["model_year"]           = details.get("Model Year", None)
    car["steering"]             = details.get("Steering", None)
    car["level"]                = details.get("Level", None)
    car["car_body_color"]       = details.get("Car body color", None)
    car["seat_count"]           = details.get("Seat Count", None)
    car["engine"]               = details.get("Engine", None)
    car["drive_mode"]           = details.get("Drive Mode", None)
    car["motor_power_kw"]       = details.get("Motor Power(kW)", None)
    car["battery_capacity_kwh"] = details.get("Battery Capacity(kWh)", None)
    car["door_count"]           = details.get("Door Count", None)

    return car


def save_data(data, filename="cars.csv"):
    df = pd.DataFrame(data)
    df.to_csv(filename, index=False)
    print(f"Saved {len(data)} cars to {filename}")


try:
    while True:
        wait.until(EC.presence_of_all_elements_located((By.CLASS_NAME, "item_car")))
        car_cards = driver.find_elements(By.CLASS_NAME, "item_car")

        for card in car_cards:
            try:
                car = get_car_details(card)
                all_cars.append(car)
            except:
                continue
        print(f"Total scraped: {len(all_cars)}")

        # autosave every 20.000 cars
        if len(all_cars)%20000 == 0 and len(all_cars)>0:
            save_data(all_cars, "cars_autosave.csv")

        # checking last page
        next_btn = driver.find_element(By.CLASS_NAME, "btn-next")
        if next_btn.get_attribute("disabled"):
            print("Last page reached!")
            break

        next_btn.click()
        time.sleep(2)

except KeyboardInterrupt:
    print("\n Stopped by user!")

except Exception as e:
    print(f"\n Error: {e}")

finally:
    save_data(all_cars, "cars_final.csv")
    driver.quit()
    print("Done!")
```
