```python
try:
    df_clean=preprocessing.handle_missing_values()
    logging.info(f"tushurib qoldirilgan qiymatlar toldirildi {df_clean.isnull().sum()}")
except Exception as e:
    logging.error(f"xatolik bor")
```

```python
try:
    df_free_dupicated=preprocessing.remove_duplicates()
    logging.info(f"duplicated bolgan elementlar tashlab yuborildi {df_free_dupicated.info()}")
except Exception as e:
    logging.error(f"xatolik bor")
```

```python
try:
    df_encoded=preprocessing.encode_categorical()
    logging.info(f"datasetimiz encoding qilindi {df_encoded.info()}")
except Exception as e:
    logging.error(f"xatolik bor")
```

```python
try:
    df_scaled=preprocessing.scale_numeric()
    logging.info(f"datasetimiz scaling qilindi {df_scaled.head()}")
  
except Exception as e:
    logging.error(f"xatolik bor")
```
