# **Simple pipeline tasks**

### 1

```python
# writing a function that doubles given integer number.

def double(num):
    return num*2

# writing a function that adds 100 to the given number.

def add_100(num):
    return num+100

# writing a function that divides given number by 2.

def divide(num):
    return int(num/2)

# writing a function that squares given number by 2. 

def square(num):
    return num**2

pipeline = [double, add_100, divide, square]
data = 500
for step in pipeline:
    data = step(data)
print(data)

```

```python
square(divide(add_100(double(500))))
```

### 2

```python
# writing a function that capitalize the first character of the string.

def capitalize(str):
    return str.capitalize()

# writing a function that reverses given string.
def rewerse(str):
    return str[::-1]

# writing a function that returns the length of given string.

def length(str):
    return len(str)

pipeline_2 = [capitalize, rewerse, length]
data = "Otabek Razzoqov"
for step in pipeline_2:
    data = step(data)
print(data)
```

```python
# writing a function that checks whether given card number is valid.

def valid_card(num):
    if len(str(num)) == 16:
        return num
    else:
        return f"invalid card"
    

# writing a function that checks whether the card is HUMO or UZCARD.

def humo_or_uzcard(num):
    if str(num).startswith("5614") and len(str(num))==16:
        return num
    else:
        return f"Kartangiz UZCARD emas!"

humo_or_uzcard(5614123412341234)

# writing a function that checks if the card owner is eligible for loan or not.

def loan_eligible(num):
    if humo_or_uzcard(num) == num:
        return f"You are eligible for loan"
    else:
        f"Your card is not UZCARD"


pipeline_3 = [valid_card, humo_or_uzcard, loan_eligible]
data = 5614101029293838
for step in pipeline_3:
    data = step(data)
print(data)
```

### 4

```python
# writing a function that strips given string.

def remove_spaces(text):
    return text.strip()

# writing a function that turns the first char of the str into lower.

def make_lowercase(text):
    return text.lower()

# writing a function that replaces " " with "-".

def replace_spaces_with_dash(text):
    return text.replace(" ", "-")

# wriring a function that adds "ev" prefix to the given str.

def add_prefix(text):
    return "ev-" + text


def run_pipeline(value, functions):
    for function in functions:
        value = function(value)
    return value



pipeline_4 = [remove_spaces, make_lowercase, replace_spaces_with_dash, add_prefix]

result = run_pipeline("  Tesla Model 3  ", pipeline_4)

print(result)
```
