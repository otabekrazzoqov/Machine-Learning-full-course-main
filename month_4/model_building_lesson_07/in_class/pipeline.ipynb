```python
# Istalgan text kiritilsin:
#1 lowercase(kichkina xarflarda bajarsin)
#2 orasidagi empty spacelarni tozalasin
#3 teskaridan yozib bersin
```

```python
#1 lowercase(kichkina xarflarda bajarsin)
def to_lower(text):
    return text.lower()
```

```python
#2 orasidagi empty spacelarni tozalasin
def remove_space(text):
    return text.replace(" ","")
```

```python
#3 teskaridan yozib bersin
def teskari(text):
    return text[::-1]
```

```python
pipeline=[to_lower,remove_space,teskari]
data='Hello AI'
for step in pipeline:
    data=step(data)
print(data)
```
