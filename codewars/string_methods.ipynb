```python
"hello".upper()          # → "HELLO"
"HELLO".lower()          # → "hello"
"hello world".title()    # → "Hello World"
"Hello".swapcase()       # → "hELLO"
"hello".capitalize()     # → "Hello"


"hello".find("l")        # → 2 (first index, -1 if not found)
"hello".index("l")       # → 2 (raises error if not found)
"hello".count("l")       # → 2
"hello".startswith("he") # → True
"hello".endswith("lo")   # → True
"hello".replace("l","r") # → "herro"
"hello" in "hello world" # → True

"  hello  ".strip()      # → "hello"
"  hello  ".lstrip()     # → "hello  "
"  hello  ".rstrip()     # → "  hello"
"xxhelloxx".strip("x")   # → "hello"

"a,b,c".split(",")       # → ["a", "b", "c"]
"hello world".split()    # → ["hello", "world"]
",".join(["a","b","c"])  # → "a,b,c"
" ".join(["hello","world"]) # → "hello world"

# f-string (most common)
name = "Otabek"
f"Hello {name}"          # → "Hello Otabek"

# format()
"Hello {}".format(name)  # → "Hello Otabek"

# padding
"hello".center(11)       # → "   hello   "
"hello".ljust(10)        # → "hello     "
"hello".rjust(10)        # → "     hello"
"42".zfill(5)            # → "00042"

"hello".isalpha()        # → True  (only letters)
"123".isdigit()          # → True  (only digits)
"hello123".isalnum()     # → True  (letters + digits)
"   ".isspace()          # → True  (only whitespace)
"HELLO".isupper()        # → True
"hello".islower()        # → True


s = "hello world"
s[0:5]                   # → "hello"
s[-5:]                   # → "world"
s[::-1]                  # → "dlrow olleh" (reverse)
s[::2]                   # → "hlowrd" (every 2nd char)
```
